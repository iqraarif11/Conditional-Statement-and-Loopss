# Loops
## For loop

```python
for i in range(1,10):
    print(i) 
```

## While loop

```python
i = 1
while i <= 7:
    print(i)
    i += 1
```

```python
i = 10
while i >= 3:
    print(i)
    i -= 1
```

## Nested loop

```python
for i in range(1,5):
    for j in range(1,4):
        print(f"i={i}, j={j}")
```

# break and continue in loop
## break

```python
for i in range (1,7):
    if i == 5:
        break
    print(i)
```

## continue

```python
for i in range(2,10):
    if i == 5:
         continue
    print(i)
```
